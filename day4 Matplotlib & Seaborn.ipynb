```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,10,100)
```

```python
y = np.sin(x)
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
ax.plot(x,y, label = "sinx(x)", color="Gold")
ax.set_xlabel("x")
ax.set_title("sine_wave")
ax.legend()
plt.show()
```

```python
np.random.seed(42)
```

```python
epochs = np.arange(1,21)
train_loss = np.exp(-epochs/5) + np.random.rand(20)*0.5
```

```python
val_loss = np.exp(-epochs/6) + np.random.rand(20)*0.08
```

```python
plt.plot(epochs, train_loss, label = "train")
plt.plot(epochs, val_loss, label = "val")
plt.xlabel("epochs")
plt.ylabel("loss")
plt.title("training vs validation loss")
plt.legend()
plt.show()
```

```python
## Histogram

data = np.random.randn(1000)
plt.hist(data, bins = 30, color = "silver", edgecolor ="green")
plt.xlabel("value")
plt.ylabel("frequency")
plt.title("Histogram")
plt.show()
```

```python
## Subplot

fig, axes = plt.subplots(1,2,figsize = (10,4))
axes[0].plot(epochs, train_loss)
axes[1].hist(data,bins = 30)
```
